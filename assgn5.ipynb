```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("Housing.csv")
data.head()
```

```python
# Q1.  Write a program to find the average price of houses grouped by the number of bedrooms. 
# Display the result in descending order of average price.

data.groupby("bedrooms")["price"].mean(numeric_only=True).sort_values(ascending=False)
```

```python
#Q2. Create a program to calculate the total area of houses grouped by stories. 
# Display only the top 2 with the largest total area.
data.groupby("stories")["area"].sum(numeric_only=True).sort_values(ascending=False).head(2)
```

```python
#Q3.  Determine the maximum price of houses based on whether they have hotwaterheating or not. 
# Also display the data of both rows. 

data.groupby("hotwaterheating")["price"].max(numeric_only=True)
```

```python
#Q4.  Add a column named PricePerSqFt to calculate the price per square foot for each house. 
# Group the dataset by the number of floors and find the average PricePerSqFt for each group.

data["PricePerSqFt"] = data["price"] / data["area"]
data.groupby("stories")["PricePerSqFt"].mean(numeric_only=True)
```

```python
#Q5. Group the data by parking and count the total number of houses in each number of parking. Also sort in acsending order.
data.groupby("parking").size().sort_values(ascending=True)
```

```python
#Q6.  Write a program to group houses based on the number of bathrooms and find the minimum, maximum, and average area (square feet) for each bathroom group.  

data.groupby("bathrooms")["area"].agg(["min", "max", "mean"], numeric_only=True)
```

```python
#Q7. Perform all the attributes and methods of pandas groupby object on this dataset which are discussed in class and analyse the data.
grouped = data.groupby("bedrooms")
print(grouped.size())
```

```python
#Q8. Write a python program which display the name and data of the individual groups and performing groupby on furnishingstatus column of the dataset.

grouped_furnishing = data.groupby("furnishingstatus")
for name, group in grouped_furnishing:
    print(f"Group: {name}")
    print(group)
```
